```python
import pandas as pd

df = pd.read_csv("student-mat.csv",sep=";" )
df.head()
```

```python
df["reason"]
```

```python
 # Column 1 School Replace short form to there actual name & make it Clean Struture
df["school"] = df["school"].str.replace({"GP" : "Gabriel Pereira", "MS" :"Mousinho da Silveira"}).str.strip().str.title()
#df["school"]

# Column 2 sex Replace Short form to there actual name & make it clean structure
df["sex"] = df["sex"].str.replace({"F":"Female", "M":"Male"}).str.strip().str.title()
#df["sex"]

# Column 3 Remove Extra Spaces
df["age"] = df["age"].str.strip() 
#df["age"]

# Column 4 address Replace Short from to there actual name & make it clean struture
df["address"] = df["address"].str.replace({"U":"urban", "R":"rural"}).str.strip().str.capitalize()
#df["address"]

 # Column 5 Family_Size Replace short form to there actual name & make it Clean Struture
df["famsize"] = df["famsize"].str.replace({"GT3":"greater than 3", "LE3":"less or equal to 3"}).str.strip().str.title()
#df["famsize"]

# Column 6 Pstatus - parent's cohabitation status
df["Pstatus"] = df["Pstatus"].str.replace({"A" : "apart", "T":"living together" })
#df["Pstatus"]

# Column 7 Medu - mother's education (numeric: 0 - none,  1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
df["Medu"] = df["Medu"].replace({0:"none", 1:"primary education (4th grade)", 2:"5th to 9th grade", 3:"secondary education", 4:"higher education" })
#df["Medu"].unique()


# Column 8 Fedu - father's education (numeric: 0 - none,  1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
df["Fedu"] = df["Fedu"].replace({0:"none", 1:"primary education (4th grade)", 2:"5th to 9th grade", 3:"secondary education", 4:"higher education" })
#df["Fedu"]

# Column 9 Mjob - mother's job (nominal: "teacher", "health" care related, civil "services" (e.g. administrative or police), "at_home" or "other")
df["Mjob"] = df["Mjob"].str.strip()
#df["Mjob"]

# Column 10  Fjob - father's job (nominal: "teacher", "health" care related, civil "services" (e.g. administrative or police), "at_home" or "other")
df["Fjob"] = df["Fjob"].str.strip()
#df["Fjob"]

```
