```python
import pandas as pd

df = pd.read_csv("student-por.csv",sep=";")

df.head()
```

```python
# ============================================================
# PREPROCESSING - PORTUGUESE LANGUAGE COURSE
# student-por.csv
# ============================================================

# Column 1: school
# Replace short form with actual school name + clean structure
df["school"] = (
    df["school"]
    .replace({
        "GP": "Gabriel Pereira",
        "MS": "Mousinho da Silveira"
    })
    .astype(str)
    .str.strip()
    .str.title()
)

# Column 2: sex
# Replace short form with actual name + clean structure
df["sex"] = (
    df["sex"]
    .replace({
        "F": "Female",
        "M": "Male"
    })
    .astype(str)
    .str.strip()
    .str.title()
)

# Column 3: age
# Numeric column
df["age"] = pd.to_numeric(df["age"], errors="coerce")


# Column 4: address
# Replace short form with actual name + clean structure
df["address"] = (
    df["address"]
    .replace({
        "U": "Urban",
        "R": "Rural"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 5: famsize
# Family size
df["famsize"] = (
    df["famsize"]
    .replace({
        "GT3": "Greater Than 3",
        "LE3": "Less Or Equal To 3"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 6: Pstatus
# Parent's cohabitation status
df["Pstatus"] = (
    df["Pstatus"]
    .replace({
        "A": "Apart",
        "T": "Living Together"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 7: Medu
# Mother's education
df["Medu"] = df["Medu"].replace({
    0: "None",
    1: "Primary Education (4th Grade)",
    2: "5th To 9th Grade",
    3: "Secondary Education",
    4: "Higher Education"
})

df["Medu"] = (
    df["Medu"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 8: Fedu
# Father's education
df["Fedu"] = df["Fedu"].replace({
    0: "None",
    1: "Primary Education (4th Grade)",
    2: "5th To 9th Grade",
    3: "Secondary Education",
    4: "Higher Education"
})

df["Fedu"] = (
    df["Fedu"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 9: Mjob
# Mother's job
df["Mjob"] = (
    df["Mjob"]
    .astype(str)
    .str.strip()
    .str.replace("_", " ", regex=False)
    .str.title()
)


# Column 10: Fjob
# Father's job
df["Fjob"] = (
    df["Fjob"]
    .astype(str)
    .str.strip()
    .str.replace("_", " ", regex=False)
    .str.title()
)


# Column 11: reason
# Reason for choosing school
df["reason"] = (
    df["reason"]
    .astype(str)
    .str.strip()
    .str.replace("_", " ", regex=False)
    .str.title()
)


# Column 12: guardian
# Student's guardian
df["guardian"] = (
    df["guardian"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 13: traveltime
# Home to school travel time
df["traveltime"] = df["traveltime"].replace({
    1: "Less Than 15 Min",
    2: "15 To 30 Min",
    3: "30 Min To 1 Hour",
    4: "Greater Than 1 Hour"
})

df["traveltime"] = (
    df["traveltime"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 14: studytime
# Weekly study time
df["studytime"] = df["studytime"].replace({
    1: "Less Than 2 Hours",
    2: "2 To 5 Hours",
    3: "5 To 10 Hours",
    4: "Greater Than 10 Hours"
})

df["studytime"] = (
    df["studytime"]
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 15: failures
# Number of past class failures
df["failures"] = pd.to_numeric(
    df["failures"],
    errors="coerce"
)


# Column 16: schoolsup
# Extra educational support
df["schoolsup"] = (
    df["schoolsup"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 17: famsup
# Family educational support
df["famsup"] = (
    df["famsup"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 18: paid
# Extra paid classes within Portuguese course
df["paid"] = (
    df["paid"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 19: activities
# Extra-curricular activities
df["activities"] = (
    df["activities"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 20: nursery
# Attended nursery school
df["nursery"] = (
    df["nursery"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 21: higher
# Wants to take higher education
df["higher"] = (
    df["higher"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 22: internet
# Internet access at home
df["internet"] = (
    df["internet"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 23: romantic
# Romantic relationship
df["romantic"] = (
    df["romantic"]
    .replace({
        "yes": "Yes",
        "no": "No"
    })
    .astype(str)
    .str.strip()
    .str.title()
)


# Column 24: famrel
# Quality of family relationships
df["famrel"] = pd.to_numeric(
    df["famrel"],
    errors="coerce"
)


# Column 25: freetime
# Free time after school
df["freetime"] = pd.to_numeric(
    df["freetime"],
    errors="coerce"
)


# Column 26: goout
# Going out with friends
df["goout"] = pd.to_numeric(
    df["goout"],
    errors="coerce"
)


# Column 27: Dalc
# Workday alcohol consumption
df["Dalc"] = pd.to_numeric(
    df["Dalc"],
    errors="coerce"
)


# Column 28: Walc
# Weekend alcohol consumption
df["Walc"] = pd.to_numeric(
    df["Walc"],
    errors="coerce"
)


# Column 29: health
# Current health status
df["health"] = pd.to_numeric(
    df["health"],
    errors="coerce"
)


# Column 30: absences
# Number of school absences
df["absences"] = pd.to_numeric(
    df["absences"],
    errors="coerce"
)


# Column 31: G1
# First period Portuguese grade
df["G1"] = pd.to_numeric(
    df["G1"],
    errors="coerce"
)


# Column 32: G2
# Second period Portuguese grade
df["G2"] = pd.to_numeric(
    df["G2"],
    errors="coerce"
)


# Column 33: G3
# Final Portuguese grade - TARGET
df["G3"] = pd.to_numeric(
    df["G3"],
    errors="coerce"
)
```

```python
# Save the cleaned dataset
df.to_csv("cleaned_student_por.csv", index=False)
```
